# 1.5 Промежуточные выводы

### Сформулированные гипотезы

## 1. Гипотеза о площади (size):

Корреляция почти 0, потому что есть выбросы

Гипотеза: После удаления аномальных значений и корректного заполнения пропусков, площадь станет основным положительным предиктором цены (чем больше площадь, тем выше цена).

## 2. Гипотеза о локации (address):

В данных присутствуют разные районы (например, престижный Beşiktaş/Levent и более удаленный Çorlu).
Гипотеза: Район расположения (извлеченный из столбца address) оказывает сильное влияние на стоимость. Объекты в центральных или деловых районах будут иметь статистически значимую премию к цене по сравнению с пригородами, даже при одинаковой площади.

## 3. Гипотеза о возрасте здания (building_age):

Данные содержат категории от "0" (новостройка) до "40 ve üzeri" (40 лет и более).
Гипотеза: Новые здания (возраст 0–5 лет) будут стоить в среднем дороже, чем старые ("40 ve üzeri"), из-за современных стандартов строительства и отсутствия необходимости в капитальном ремонте. Зависимость, вероятно, будет нелинейной (резкое падение цены после первых 5-10 лет, затем стабилизация).

## 4. Гипотеза о планировке (room_count):

Гипотеза: Количество комнат имеет положительную корреляцию с ценой, но прирост стоимости нелинеен. Переход от студии (1+0) к стандартной квартире (2+1) дает больший абсолютный прирост цены, чем переход от 3+1 к 4+1, так как спрос на массовом рынке сосредоточен именно на 1+1 и 2+1.

# 2.1 РАБОТА С ПРОПУСКАМИ

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.impute import KNNImputer
from sklearn.preprocessing import OrdinalEncoder, OneHotEncoder, RobustScaler, StandardScaler, MinMaxScaler
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from statsmodels.stats.outliers_influence import variance_inflation_factor
import warnings
warnings.filterwarnings('ignore')

df = pd.read_csv('real_estate_data.csv', low_memory=False)
print("Исходная размерность:", df.shape)
print("2.1 РАБОТА С ПРОПУСКАМИ")

# Удаляем столбцы
df = df.drop(columns=['furnished', 'end_date'])

# Удаляем строки с пропуском в цене (и валюте)
df = df.dropna(subset=['price', 'price_currency'])

# Заполняем size медианой по room_count (группировка)
df['size'] = df.groupby('room_count')['size'].transform(
    lambda x: x.fillna(x.median())
)
# Если в какой-то группе медиана NaN (мало данных) — общая медиана
df['size'] = df['size'].fillna(df['size'].median())

# Заполняем категориальные модой
for col in ['building_age', 'total_floor_count', 'floor_no', 'heating_type']:
    df[col] = df[col].fillna(df[col].mode()[0])

print("После обработки пропусков:", df.shape)
print("Остаточные пропуски:", df.isnull().sum().sum())
```

# 2.2 ОБРАБОТКА ВЫБРОСОВ

```python
print("2.2 ОБРАБОТКА ВЫБРОСОВ")

# Удаляем отрицательные цены
df = df[df['price'] > 0]

# Винзоризация цены
price_upper = df['price'].quantile(0.99)
df = df[df['price'] <= price_upper]
print(f"Удалены объекты с ценой выше {price_upper:,.0f} TRY (99-й перцентиль)")

# Винзоризация площади
size_upper = df['size'].quantile(0.99)
df = df[df['size'] <= size_upper]
print(f"Удалены объекты с площадью выше {size_upper:,.0f} м² (99-й перцентиль)")

print(f"Размерность после обработки выбросов: {df.shape}")

# Создаем price_log
df['price_log'] = np.log1p(df['price'])

# Визуализация
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].hist(df['price'], bins=50)
axes[0].set_xscale('log')
axes[0].set_title('Распределение цены (лог-шкала)')
axes[0].set_xlabel('Цена (TRY)')

axes[1].hist(df['price_log'], bins=50)
axes[1].set_title('Распределение log1p(price)')
axes[1].set_xlabel('log1p(price)')

plt.tight_layout()
plt.show()
```

# 2.3 КОДИРОВАНИЕ КАТЕГОРИАЛЬНЫХ ПРИЗНАКОВ

```python
print("2.3 КОДИРОВАНИЕ КАТЕГОРИАЛЬНЫХ ПРИЗНАКОВ")

# 1. Извлекаем район из address
df['district'] = df['address'].apply(
    lambda x: x.split('/')[1].strip() if pd.notna(x) and len(x.split('/')) > 1 else 'Unknown'
)
df = df.drop(columns=['address'])

print(f"Всего уникальных районов (до группировки): {df['district'].nunique()}")
print("Примеры районов:", df['district'].value_counts().head(10).index.tolist())

# 2. Группируем редкие районы (< 0.5%) в "Другое"
threshold = 0.005
district_counts = df['district'].value_counts(normalize=True)
rare_districts = district_counts[district_counts < threshold].index
df['district'] = df['district'].replace(rare_districts, 'Другое')

print(f"Всего уникальных районов (после группировки): {df['district'].nunique()}")
print("Топ-10 районов:", df['district'].value_counts().head(10).index.tolist())

# 3. Порядковое кодирование room_count
room_order = [['1+0', '1+1', '2+1', '2+2', '3+1', '3+2', '4+1', '4+2', '5+1', '5+2', '+']]
existing_rooms = [r for r in room_order[0] if r in df['room_count'].unique()]
room_order = [existing_rooms]

room_encoder = OrdinalEncoder(categories=room_order, handle_unknown='use_encoded_value', unknown_value=-1)
df['room_count_enc'] = room_encoder.fit_transform(df[['room_count']])

# 4. Порядковое кодирование building_age
def parse_building_age(x):
    if pd.isna(x):
        return np.nan
    x = str(x)
    if x == '0':
        return 0
    if 've üzeri' in x:
        num = int(x.split()[0])
        return num + 5
    if '-' in x:
        parts = x.split('-')
        low = int(parts[0])
        high = int(parts[1].split()[0])
        return (low + high) / 2
    try:
        return float(x)
    except:
        return np.nan

df['building_age_num'] = df['building_age'].apply(parse_building_age)
df['building_age_num'] = df['building_age_num'].fillna(df['building_age_num'].median())

# 5. Порядковое кодирование этажей
def parse_floor(x):
    if pd.isna(x):
        return np.nan
    x = str(x)
    if 've üzeri' in x:
        return int(x.split()[0]) + 5
    if 'Yüksek Giriş' in x or 'Zemin' in x:
        return 0
    try:
        return float(x)
    except:
        return np.nan

df['floor_no_num'] = df['floor_no'].apply(parse_floor)
df['floor_no_num'] = df['floor_no_num'].fillna(df['floor_no_num'].median())

df['total_floor_num'] = df['total_floor_count'].apply(parse_floor)
df['total_floor_num'] = df['total_floor_num'].fillna(df['total_floor_num'].median())

# 6. One-Hot Encoding для номинальных признаков
nominal_cols = ['type', 'sub_type', 'heating_type', 'price_currency', 'district']
df = pd.get_dummies(df, columns=nominal_cols, prefix=nominal_cols, drop_first=False)

# 7. Создаем признак is_central до удаления OHE-колонок района
central_districts = ['Beşiktaş', 'Şişli', 'Kadıköy', 'Bakırköy', 'Ataşehir', 'Levent']
central_cols = [c for c in df.columns if any(d in c for d in central_districts)]

if len(central_cols) > 0:
    df['is_central'] = df[central_cols].sum(axis=1).clip(upper=1)
else:
    df['is_central'] = 0

print(f"\nРаспределение признака is_central:")
print(df['is_central'].value_counts())
print(f"Процент центральных районов: {df['is_central'].mean() * 100:.2f}%")


df['price_per_m2'] = df['price'] / (df['size'] + 1)
df['age_cat_new'] = (df['building_age_num'] <= 5).astype(int)
df['age_cat_old'] = (df['building_age_num'] >= 25).astype(int)

# Удаляем исходные категориальные столбцы
df = df.drop(columns=['room_count', 'building_age', 'total_floor_count', 'floor_no'], errors='ignore')

print(f"\nРазмерность после кодирования: {df.shape}")
print("Пример OHE-колонок:", [c for c in df.columns if c.startswith('district_')][:5])
```

# 2.4 УДАЛЕНИЕ МУЛЬТИКОЛЛИНЕАРНЫХ ПРИЗНАКОВ (VIF)

```python
print("2.4 УДАЛЕНИЕ МУЛЬТИКОЛЛИНЕАРНЫХ ПРИЗНАКОВ (VIF)")

numeric_for_vif = df.select_dtypes(include=[np.number]).drop(
    columns=['price', 'price_log', 'id'], errors='ignore'
)

vif_data = pd.DataFrame()
vif_data['feature'] = numeric_for_vif.columns
vif_data['VIF'] = [variance_inflation_factor(numeric_for_vif.values, i)
                   for i in range(numeric_for_vif.shape[1])]

print("\nVIF до удаления:")
print(vif_data.sort_values('VIF', ascending=False).head(10))

# Удаляем признаки с VIF > 10
high_vif = vif_data[vif_data['VIF'] > 10]['feature'].tolist()
print(f"\nПризнаки с VIF > 10 (будут удалены): {high_vif}")

df = df.drop(columns=[c for c in high_vif if c in df.columns], errors='ignore')
print(f"Размерность после удаления мультиколлинеарных: {df.shape}")
```

# 3.3 Обучение моделей с подбором гиперпараметров

```python
import numpy as np
import pandas as pd
import time

from sklearn.linear_model import Ridge
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.model_selection import RandomizedSearchCV


print("\n3.3 Обучение моделей")

best_models_sale = {}
best_models_rent = {}


SUB = 25000
np.random.seed(42)

idx_s = np.random.choice(len(X_train_sale), size=min(SUB, len(X_train_sale)), replace=False)
idx_r = np.random.choice(len(X_train_rent), size=min(SUB, len(X_train_rent)), replace=False)

Xs_tr, ys_tr = X_train_sale.iloc[idx_s], y_train_sale.iloc[idx_s]
Xr_tr, yr_tr = X_train_rent.iloc[idx_r], y_train_rent.iloc[idx_r]

print(f"Поиск на подвыборке: sale={Xs_tr.shape}, rent={Xr_tr.shape}")
print(f"Финал на полном train: sale={X_train_sale.shape}, rent={X_train_rent.shape}\n")


models_to_tune = [
    ('Ridge', Ridge(random_state=42),
     {'alpha': [0.1, 1.0, 10.0]}),
    ('RandomForest', RandomForestRegressor(random_state=42, n_jobs=1),
     {'n_estimators': [80, 120],
      'max_depth': [10, 14],
      'min_samples_leaf': [3, 6]}),
    ('GradientBoosting', GradientBoostingRegressor(random_state=42),
     {'n_estimators': [80, 120],
      'learning_rate': [0.08, 0.12],
      'max_depth': [3, 4],
      'subsample': [0.8, 1.0]}),
]

def tune_and_fit(name, model, params, X_sub, y_sub, X_full, y_full):
    t0 = time.time()

    search = RandomizedSearchCV(
        estimator=model, param_distributions=params,
        n_iter=6, cv=2,
        scoring='neg_mean_absolute_error',
        random_state=42, n_jobs=-1
    )
    search.fit(X_sub, y_sub)
    t_search = time.time() - t0
    print(f"  [{name}] поиск: {t_search:6.1f}s | {search.best_params_} | MAE(cv)={-search.best_score_:.4f}")


    t1 = time.time()
    best = search.best_estimator_
    best.fit(X_full, y_full)
    t_fit = time.time() - t1
    print(f"  [{name}] финал на полном train: {t_fit:6.1f}s")
    return best


print("=" * 70)
print("ОБУЧЕНИЕ ДЛЯ ПРОДАЖИ (listing_type=1)")
print("=" * 70)
for name, model, params in models_to_tune:
    best_models_sale[name] = tune_and_fit(name, model, params, Xs_tr, ys_tr, X_train_sale, y_train_sale)
print("Продажа готова\n")


print("=" * 70)
print("ОБУЧЕНИЕ ДЛЯ АРЕНДЫ (listing_type=2)")
print("=" * 70)
for name, model, params in models_to_tune:
    best_models_rent[name] = tune_and_fit(name, model, params, Xr_tr, yr_tr, X_train_rent, y_train_rent)
print("Аренда готова")
```

# 3.4 Оценка моделей на тестовой выборке

```python
import numpy as np
import pandas as pd
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import matplotlib.pyplot as plt

print("\n3.4 Оценка моделей на тестовой выборке")

def evaluate_model(model, X_test, y_test):
    y_pred_log = model.predict(X_test)
    y_true = np.expm1(y_test)
    y_pred = np.expm1(y_pred_log)
    
    mae = mean_absolute_error(y_true, y_pred)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    r2 = r2_score(y_true, y_pred)
    
    epsilon = 1e-5
    mape = np.mean(np.abs((y_true - y_pred) / (y_true + epsilon))) * 100
    
    return {'MAE': mae, 'RMSE': rmse, 'R2': r2, 'MAPE (%)': mape}

print("\n" + "="*70)
print("РЕЗУЛЬТАТЫ ДЛЯ ПРОДАЖИ")
print("="*70)



plot_sale_data = []
for name, model in best_models_sale.items():
    print(f"Оценка модели: {name} (Продажа)...")
    metrics = evaluate_model(model, X_test_sale, y_test_sale)
    metrics['Model'] = name
    plot_sale_data.append(metrics)

plot_sale = pd.DataFrame(plot_sale_data)
plot_sale['MAE'] = plot_sale['MAE'].map('{:,.0f}'.format)
plot_sale['RMSE'] = plot_sale['RMSE'].map('{:,.0f}'.format)
plot_sale['MAPE (%)'] = plot_sale['MAPE (%)'].map('{:.2f}'.format)
plot_sale['R2'] = plot_sale['R2'].map('{:.4f}'.format)
plot_sale = plot_sale[['Model', 'R2', 'MAE', 'RMSE', 'MAPE (%)']]
print(plot_sale.to_string(index=False))

#  АРЕНДА
print("\n" + "="*70)
print("РЕЗУЛЬТАТЫ ДЛЯ АРЕНДЫ")
print("="*70)



plot_rent_data = []
for name, model in best_models_rent.items():
    print(f"Оценка модели: {name} (Аренда)...")
    metrics = evaluate_model(model, X_test_rent, y_test_rent)
    metrics['Model'] = name
    plot_rent_data.append(metrics)

plot_rent = pd.DataFrame(plot_rent_data)
plot_rent['MAE'] = plot_rent['MAE'].map('{:,.0f}'.format)
plot_rent['RMSE'] = plot_rent['RMSE'].map('{:,.0f}'.format)
plot_rent['MAPE (%)'] = plot_rent['MAPE (%)'].map('{:.2f}'.format)
plot_rent['R2'] = plot_rent['R2'].map('{:.4f}'.format)
plot_rent = plot_rent[['Model', 'R2', 'MAE', 'RMSE', 'MAPE (%)']]
print(plot_rent.to_string(index=False))

#  ВИЗУАЛИЗАЦИЯ СРАВНЕНИЯ
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

# R2 для продажи
r2_vals_sale = [float(x) for x in plot_sale['R2']]
axes[0, 0].bar(plot_sale['Model'], r2_vals_sale, color='teal')
axes[0, 0].set_title('R² Score — ПРОДАЖА')
axes[0, 0].set_ylim(0, max(r2_vals_sale) * 1.2 if max(r2_vals_sale) > 0 else 1)
axes[0, 0].set_ylabel('R²')

# MAPE для продажи
mape_vals_sale = [float(x) for x in plot_sale['MAPE (%)']]
axes[0, 1].bar(plot_sale['Model'], mape_vals_sale, color='coral')
axes[0, 1].set_title('MAPE (%) — ПРОДАЖА')
axes[0, 1].set_ylabel('MAPE (%)')

# R2 для аренды
r2_vals_rent = [float(x) for x in plot_rent['R2']]
axes[1, 0].bar(plot_rent['Model'], r2_vals_rent, color='steelblue')
axes[1, 0].set_title('R² Score — АРЕНДА')
axes[1, 0].set_ylim(0, max(r2_vals_rent) * 1.2 if max(r2_vals_rent) > 0 else 1)
axes[1, 0].set_ylabel('R²')

# MAPE для аренды
mape_vals_rent = [float(x) for x in plot_rent['MAPE (%)']]
axes[1, 1].bar(plot_rent['Model'], mape_vals_rent, color='gold')
axes[1, 1].set_title('MAPE (%) — АРЕНДА')
axes[1, 1].set_ylabel('MAPE (%)')

plt.tight_layout()
plt.show()

print("\nБлок 3.4 завершён успешно.")
```
